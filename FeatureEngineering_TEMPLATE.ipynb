```python
raw_url = filepath+"/images/mse_logo.png"
#display(Image(url=raw_url))
response = requests.get(raw_url)
# Check if the request was successful
if response.status_code == 200:
    # 2. Get the bytes directly from the response (no need to save to a file)
    image_bytes = response.content
    # 3. Encode to base64
    encoded_string = base64.b64encode(image_bytes).decode('utf-8')
else:
    print("Failed to download image")
# 4. Embed it in the HTML <img> tag
html_header = f"""
<img src="data:image/png;base64,{encoded_string}" width="400" align="right"/>
<div style="text-align: left"> <b> Machine Learning FTP_MachLe </b> <br> SPR 2026 <br> <a href="mailto:christoph.wuersch@ost.ch"> Christoph Würsch </a> </div>
"""
```
